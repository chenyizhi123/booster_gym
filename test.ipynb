```python
# 首先安装rsl_rl模块
import subprocess
import sys
import os

print("📦 安装rsl_rl模块...")

# 获取当前工作目录
current_dir = os.getcwd()
rsl_rl_dir = os.path.join(current_dir, "rsl_rl")

# 使用pip安装本地rsl_rl模块
try:
    result = subprocess.run([
        sys.executable, "-m", "pip", "install", "-e", rsl_rl_dir
    ], capture_output=True, text=True, cwd=current_dir)
    
    if result.returncode == 0:
        print("✅ rsl_rl模块安装成功")
        print(result.stdout)
    else:
        print("❌ rsl_rl模块安装失败")
        print("STDOUT:", result.stdout)
        print("STDERR:", result.stderr)
        
except Exception as e:
    print(f"❌ 安装过程中出错: {e}")

print("\n" + "="*50)
```

```python
import joblib
import numpy as np  # 用于保存数组到txt

a = 'data//0-rub055_0025_kicking1_poses.pkl'
data = joblib.load(a)

# 获取pose_aa的第一帧数据（shape为(1, 31, 3)）
pose_aa_first_frame = data['0-rub055_0025_kicking1_poses']['root_rot']

# 保存到txt文件
# 由于是三维数组，可通过reshape转为二维方便阅读（保持31行3列的结构）
np.savetxt(
    'root_rot.txt',  # 保存的文件名
    pose_aa_first_frame,  # 转为(31, 3)的二维数组
    fmt='%.6f',  # 保留6位小数
    header='root_rot数据，每个的轴角向量(x, y, z)）',  # 文件头部说明
    comments=''  # 去除注释符'#'，让header正常显示
)

print("数据已保存到 root_rot.txt")
```
